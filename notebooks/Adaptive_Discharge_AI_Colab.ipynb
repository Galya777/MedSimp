```
# Setup: install dependencies and prepare environment
import sys, subprocess, os, textwrap
from pathlib import Path

subprocess.check_call([sys.executable, "-m", "pip", "install", "-q",
                       "torch==2.3.1", "transformers==4.48.0", "accelerate==0.34.2",
                       "bitsandbytes==0.43.1", "textstat==0.7.4", "readability-lxml==0.8.1",
                       "sentencepiece==0.2.0", "safetensors==0.4.3"])  # streamlit not required here

# Ensure repo modules are importable when running from repo root
repo_root = Path.cwd()
if str(repo_root / 'add') not in sys.path:
    sys.path.append(str(repo_root / 'add'))

os.environ.setdefault('MODEL_ID', 'google/medgemma-4b-it')

from add.inference import load_model, generate_response
from add.structure import extract_sections
from add.litteracy import readability_scores
```

```
# Demo: end-to-end example
tokenizer, model = load_model()

discharge_text = textwrap.dedent('''
- Take 10 mg lisinopril every morning.
- Avoid driving for 24 hours after sedation.
- Call 911 if you have chest pain or difficulty breathing.
- Follow up with cardiology clinic in 7 days.
''').strip()

sections = extract_sections(discharge_text)
out = generate_response(tokenizer, model, sections, profile='low_literacy')
print(out)
print('\nReadability FKGL before/after:', readability_scores(discharge_text)['fkgl'], '→', readability_scores(out)['fkgl'])
```
